```python

# --- Test Data for Nest method ---

data = {'a': 1, 'b': {'c': 'foo', 'd': 'bar', 'e': 4}, 'f': [1, 2, 3]}
data_tester = {'a': 1, 'b': {'c': 'foo', 'd': 'bar', 'e': 4}, 'f': [{'k':"r"}, {"l": 2}, 3]}
data1 = {"a": 2, "b": {"c": 7, "d": 'bar', "e": 4}, "f": [1, 2, 3]}
data2 = {"a": 2, "b": {"c": 'bar'}, "g": 7}
data3 = [
    { "a": 1, "common": { "x": 10, "b": {"c": 7} }, "u1": "foo" },
    { "a": 2, "common": { "x": "twenty" }, "u2": "bar" },
    { "a": 3, "common": { "x": 30 } }
]
data4 = [
    { "a": 100, "common": { "x": 40 }, "v1": 3.14 },
    { "a": 200, "common": { "x": 20 }},
    { "a": 300, "common": { "x": "gamma" }, "v2": 2.71 }
]
```

```python
# Testing new Nest class
from DataGuidePathClass import DataGuidePath, Path
print("--- Example 1: Group by 'a' and 'b.c' ---")

# 1. Initialize DataGuidePath
dg1 = DataGuidePath()

# 2. Insert all data
dg1.insert_document(data)
dg1.insert_document(data_tester)
dg1.insert_document(data1)
dg1.insert_document(data2)
# This is a list of documents
dg1.insert_document(data3)
# This is a list of documents
dg1.insert_document(data4) 

print(f"Total documents in original guide: {dg1.total_docs}\n")
print("Original Data Guide:")
dg1.print_guide()

```

```python
# 3. Calling group by ('a', 'b.c'). 
grouping_keys_ex1 = [Path('a'), Path('b.c')] 
new_nested_key_ex1 = 'grouped_data_fuzzy'

# new_dg1 will store the new DataGuide object returned by "nest_by_grouping_keys "
# report1 stores returns a dictionary of metrics and estimations
# "grouping_keys_ex1" is a list of define the criterai for nesting, more than 2 can be passed
# "new_nested_key_ex1" name of the new key under which the the identified grouping paths will be nested
# "include_partial_or_null=False" this is a boolean flag that controls how paths not matching any of the grouping_keys_ex1 are handled in the transformed schema.
new_dg1, report1 = dg1.nest_by_grouping_keys(grouping_keys_ex1, new_nested_key_ex1, include_partial_or_null=False)

print("\nTransformed Data Guide (Example 1):")
new_dg1.print_guide()
```

```python
print("\nTransformed Data Guide (Example 1):")
new_dg1.print_guide()
```

```python
print("\nReport on Impacted Documents (Example 1):")
print(f"\tTotal documents in original guide: {report1['total_documents_in_guide']}")
print(f"\tMax documents estimated to be impacted (all documents): {report1['max_impacted_documents_estimate']}")
print(f"\tMin documents estimated to have ALL EXACT grouping keys ('a' AND 'b.c'): {report1['min_exact_grouping_keys_present_estimate']}") 

if 'estimated_documents_with_primary_and_fuzzy_subpath' in report1:
    print(f"\tEstimated documents with ('a' AND any path containing 'b.c'): {report1['estimated_documents_with_primary_and_fuzzy_subpath']}")
print("\tGrouping key presence counts (total occurrences in guide for exact paths):")
for key, count in report1['grouping_key_presence_counts'].items():
    print(f"\t{key}: {count}")
```

```python
print("\n--- Example 2: Group by 'a' and 'common.x' ---")

# 1. Initialize DataGuidePath
dg2 = DataGuidePath()

# 2. Insert all data
dg2.insert_document(data)
dg2.insert_document(data_tester)
dg2.insert_document(data1)
dg2.insert_document(data2)
dg2.insert_document(data3)
dg2.insert_document(data4)

print(f"Total documents in original guide: {dg2.total_docs}\n")
print("Original Data Guide:")
dg2.print_guide()
```

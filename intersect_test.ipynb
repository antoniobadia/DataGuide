```python
# Docs to be used for easier test to follow
doc1 = { "sport": "soccer", "difficulty_lvl": 3, "num_balls": 1 } # original base doc
doc2 = { "sport": "dodgeball", "difficulty_lvl": 1, "num_balls": 6 } # original base doc
doc3 = { "sport": "swimming", "difficulty_lvl": 8, "involves_water": "no" } # additional key and value
doc4 = { "sport": "swimming", "difficulty_lvl": 8, "involves_water": "yes" } # additional key and value
doc5 = { "sport": "tennis", "difficulty_lvl": 5, "num_balls": "three" } # num_balls is str type instead of int
doc6 = { "sport": "soccer", "difficulty_lvl": 3, "num_balls": 1, "field_type": "grass" } # additional key and value
doc7 = { "f_name": "John", "l_name": "Smith", "age": 25} # unrelated doc
doc8 = { "sport": "dodgeball", "num_balls": 6 , "difficulty_lvl": 3} # path in different order from base doc
doc9 = { "f_name": "John", "l_name": "Smith", "age": 25, "sport": "soccer"} # different schema with 1 related path
```

```python
# Case 1: A intersection B is empty.
# Case 1.1 DG(A) and DG(B) have the same paths (maybe with different counters). 
# This means that the documents in A and B are different because they have similar 
# but different schemas (different combinations of paths) or different data values.

# 1doc each dg
dg1 = DataGuidePath()
dg1.insert_document([doc1])
dg2 = DataGuidePath()
dg2.insert_document([doc3]) 
res = dg1.intersect_new(dg2)
res.print_guide()
# since dg1 has diff paths from dg2, and vice versa, result is all 0's
```

```python
# continuing case 1.1 test but adding more documents
dg1 = DataGuidePath()
dg1.insert_document([doc1, doc2])
dg2 = DataGuidePath()
dg2.insert_document([doc3, doc4]) 
res = dg1.intersect_new(dg2)
res.print_guide()
# expect all 0's
```

```python
# case 1.1 where we have matching paths, but documents do not match
dg1 = DataGuidePath()
dg1.insert_document([doc1, doc2]) 
dg2 = DataGuidePath()
dg2.insert_document([
    { "sport": "soccer"},
    { "difficulty_lvl": 3},
    { "num_balls": 1 },
    { "sport": "dodgeball"}, 
    { "difficulty_lvl": 1},
    { "num_balls": 6 }
]) 
res = dg1.intersect_new(dg2)
res.print_guide()
# since dg's only store type and counters, the paths will match and 2 docs should be in the res
```

```python
# Case 1.2: DG(A) and DG(B) are disjoint. This means that the documents in A and B are different because they have totally different schemas.
# starting with single document
dg1 = DataGuidePath()
dg1.insert_document([doc1])
dg2 = DataGuidePath()
dg2.insert_document([doc7]) 
res = dg1.intersect_new(dg2)
res.print_guide()
# completely different schemas, so all 0's expected
```

```python
# continuing case 1.2 test but adding more documents
dg1 = DataGuidePath()
dg1.insert_document([doc1, doc2])
dg2 = DataGuidePath()
dg2.insert_document([doc7, doc9]) 
res = dg1.intersect_new(dg2)
res.print_guide()
# completely different schemas, so all 0's expected
```

```python
# Case 2: A is a subset of B -all documents in A are also in B. Hence, A intersection B is A.
# Case 2.1 DG(A) and DG(B) have the same paths. 
# This means documents in B that are not in A have similar but different schemas or different data values.
dg1 = DataGuidePath()
dg1.insert_document([doc1, doc2, doc3, doc4])
dg2 = DataGuidePath()
dg2.insert_document([doc4, doc3, doc2, doc1]) 
res = dg1.intersect_new(dg2)
res.print_guide()
# expected to have all 4 documents in res
```

```python
# Case 2.2 DG(B) has paths that are not in DG(A). 
# This means that documents in B that are not in A have different schemas from those documents in A.
dg1 = DataGuidePath()
dg1.insert_document([doc1, doc2, doc3, doc4]) # 4 docs
dg2 = DataGuidePath()
dg2.insert_document([doc6, doc5, doc4, doc3, doc2, doc1]) # 6 docs
res = dg1.intersect_new(dg2)
res.print_guide()
# Expect to find 4 matching documents at least
```

```python
#  Order of calling intersection does not matter:
dg1 = DataGuidePath()
dg1.insert_document([doc1, doc2, doc3, doc4]) # 4 docs
dg2 = DataGuidePath()
dg2.insert_document([doc6, doc5, doc4, doc3, doc2, doc1]) # 6 docs
res = dg2.intersect_new(dg1)
res.print_guide()
# expect to have at least 4 matching docs
```

```python
# Now testing the runtime for a much larger dataset
# 1.58GB of data of 1000 patients
import os
import json

# Path to folder
folder = "patient_data\sample-bulk-fhir-datasets-1000-patients\sample-bulk-fhir-datasets-1000-patients"

# Create the DataGuidePath instance
patient_data_dg = DataGuidePath()

# Loop through all .ndjson files in the folder
for filename in sorted(os.listdir(folder)):
    if not filename.endswith(".ndjson"):
        continue
    if filename == "log.ndjson":
        continue  # skip the log file

    file_path = os.path.join(folder, filename)
    print(f"Loading {file_path}...")

    # Each NDJSON file contains one JSON object per line
    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            try:
                doc = json.loads(line)
                patient_data_dg.insert_document_with_values(doc)
            except json.JSONDecodeError as e:
                print(f"Skipping invalid JSON line in {filename}: {e}")

print("\n All NDJSON files loaded into DataGuidePath (excluding log.ndjson).")

# Runtime 2m 20.5s, roughly 1min more than without these values
```

```python
# Printing the dg of patient data
patient_data_dg.print_guide_values()
# Original Counters remain the same, and distinct value counters are working well.
```

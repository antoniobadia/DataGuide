```python
# First larger test where dg is size 5-15, and dg2 is size 5-15
import random

random.seed(10)  

def make_hashable(value):
    if isinstance(value, dict):
        return frozenset((k, make_hashable(v)) for k, v in value.items())
    elif isinstance(value, list):
        return tuple(make_hashable(v) for v in value)
    else:
        return value

# since doc1–doc40 are already defined
all_docs = [globals()[f"doc{i}"] for i in range(1, 41)]

# tracking
stats = {"lower_bound": 0, "upper_bound": 0, "mid": 0}
tie_stats = {} 

num_iterations = 1000

for i in range(num_iterations):
    
    n1 = random.randint(5, 15)
    n2 = random.randint(5, 15)
    dg1_list = random.choices(all_docs, k=n1)
    dg2_list = random.choices(all_docs, k=n2)

    # convert to hashable for accurate set difference
    set1 = {make_hashable(d) for d in dg1_list}
    set2 = {make_hashable(d) for d in dg2_list}
    total_diff_docs = len(set1 - set2)

    # compute all three modes
    estimates = {}
    for mode in ["lower_bound", "upper_bound", "mid"]:
        dg_1 = DataGuidePath()
        dg_1.insert_document(dg1_list)
        dg_2 = DataGuidePath()
        dg_2.insert_document(dg2_list)
        result_diff_dg = dg_1.difference_beta(dg_2, mode=mode)
        estimates[mode] = result_diff_dg.total_docs

    # compute absolute differences
    diffs = {m: abs(estimates[m] - total_diff_docs) for m in estimates}
    min_diff = min(diffs.values())

    # find all modes that achieved the minimum distance
    tied_modes = [m for m, diff in diffs.items() if diff == min_diff]

    if len(tied_modes) == 1:
        # clear single winner
        stats[tied_modes[0]] += 1
    else:
        # record tie, but no win awarded
        tie_key = " & ".join(sorted(tied_modes))
        tie_stats[tie_key] = tie_stats.get(tie_key, 0) + 1


# --- summarize results ---
print("\n=== RESULTS AFTER", num_iterations, "RUNS ===")
for mode, count in stats.items():
    print(f"{mode:12s}: {count} unique wins")

print("\n=== TIES ===")
if tie_stats:
    for tie_combo, count in sorted(tie_stats.items(), key=lambda x: -x[1]):
        print(f"{tie_combo:25s}: {count} times")
else:
    print("No ties occurred.")
```

```python
# Docs to be used for easier test to follow
doc1 = { "sport": "soccer", "difficulty_lvl": 3, "num_balls": 1 } # original base doc
doc2 = { "sport": "dodgeball", "difficulty_lvl": 1, "num_balls": 6 } # original base doc
doc3 = { "sport": "swimming", "difficulty_lvl": 8, "involves_water": "no" } # additional key and value
doc4 = { "sport": "swimming", "difficulty_lvl": 8, "involves_water": "yes" } # additional key and value
doc5 = { "sport": "tennis", "difficulty_lvl": 5, "num_balls": "three" } # num_balls is str type instead of int
doc6 = { "sport": "soccer", "difficulty_lvl": 3, "num_balls": 1, "field_type": "grass" } # additional key and value
doc7 = { "f_name": "John", "l_name": "Smith", "age": 25} # unrelated doc
doc8 = { "sport": "dodgeball", "num_balls": 6 , "difficulty_lvl": 3} # path in different order from base doc
doc9 = { "f_name": "John", "l_name": "Smith", "age": 25, "sport": "soccer"} # different schema with 1 related path

doc11 = { "sport": "soccer", "difficulty_lvl": 3, "num_balls": [1, 2, 3] } # original base doc
doc12 = { "sport": "dodgeball", "difficulty_lvl": 1, "num_balls": 6 } # original base doc
```

```python
# Creating two dgs indentical paths and docs, 1 each
dg_1 =  DataGuidePath()
dg_1.insert_document(doc1)
dg_2 = DataGuidePath()
dg_2.insert_document(doc1)
```

```python
# showing what each dg has
print("dg1:")
dg_1.print_guide()
print("dg2:")
# showing what each dg has
dg_2.print_guide()
```

```python
# Case 1.1 Difference is Empty, identical dgs
# testing difference for two identical dgs
result_diff_dg = dg_1.difference_beta(dg_2, mode="lower_bound")
result_diff_dg.print_guide()
# works correctly, all counters are 0 and not paths since dg_1 has no paths that don't exist in dg-2
```

```python
# Case 1.2 DG(A) and DG(B) are disjoint
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2]) # no matching paths to dg2
dg_2 = DataGuidePath()
dg_2.insert_document([doc7, doc8])
result_diff_dg = dg_1.difference_beta(dg_2, mode="lower_bound")
result_diff_dg.print_guide()
# expected output is obj:1, because we are using "lower_bound", "upper_bound would yield obj:2"
```

```python
# Same test as above but in reverse call order
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2]) # no matching paths to dg2
dg_2 = DataGuidePath()
dg_2.insert_document([doc7, doc8])
result_diff_dg = dg_2.difference_beta(dg_1, mode="lower_bound")
result_diff_dg.print_guide()
# expected output is obj:1 because we are using "lower_bound", "upper_bound would yield obj:2"
```

```python
# Same test as above but using "upper_bound"
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2]) # no matching paths to dg2
dg_2 = DataGuidePath()
dg_2.insert_document([doc7, doc8])
result_diff_dg = dg_2.difference_beta(dg_1, mode="upper_bound")
result_diff_dg.print_guide()
# expected output is obj:2 because we are using "upper_bound"
```

```python
# Case 1.1 Difference is Empty cont, identical dgs, more complex
dg_1 =  DataGuidePath()
dg_1.insert_document([doc2, doc3, doc4])
dg_2 = DataGuidePath()
dg_2.insert_document([doc3, doc2, doc4])
result_diff_dg = dg_1.difference_beta(dg_2, mode="lower_bound")
result_diff_dg.print_guide()
# expected output is all 0's since dgs match, difference is all 0's
# SPECIAL CASE NEED TO MAKE THIS SHOW 1
```

```python
# Case 2.1 A is a subset of B -all documents
# new dg's with only one path different
dg_1 =  DataGuidePath()
dg_1.insert_document(doc7)
dg_2 = DataGuidePath()
dg_2.insert_document(doc9) # doc9 has one more path than doc7, all other paths the same
```

```python
# Case 2.1 A is a subset of B -all documents
# testing current difference, WORKS CORRECTLY since path d does exist in dg_2 and not in dg_1
result_diff_dg = dg_2.difference_beta(dg_1, mode="lower_bound")
result_diff_dg.print_guide()
# Works correctly, expected output is obj:1 since we have 1 doc in each dg and dg2 has 1 unique path
```

```python
# Testing the same as above but in different call order
result_diff_dg = dg_1.difference_beta(dg_2, mode="lower_bound")
result_diff_dg.print_guide()
# Works correctly, dg_1 has no unique paths, dg_2 contains all paths in dg_1, 
# Thus we expect that 0 docs will be in dg1 that are not in dg2
# Eventhough the doc in dg1 does not match the doc in dg2
# This is because it reads which docs are in dg1 but not in dg2 (or perhaps should be special case??)
```

```python
# Case 2.1 DG(A) and DG(B) have the same paths.
# Adding path to dg1 that exists in dg2, bringing total docs to 2 in dg1 and 1 doc in dg2
dg_1 =  DataGuidePath()
dg_1.insert_document(doc7)
dg_2 = DataGuidePath()
dg_2.insert_document(doc9) # contains path root.sport
dg_1.insert_document([{"sport": "tennis"}]) # adding doc with path root.sport
result_diff_dg = dg_1.difference_beta(dg_2, mode="lower_bound")
result_diff_dg.print_guide()
# The output expected is root: {'int': 0, 'str': 0, 'float': 0, 'date': 0, 'obj': 1, 'arr': 0}
# This is because we have all the same paths, but dg1 has two docs, but dg2 has one doc, so
# At least one doc in dg1 should not match dg2.
```

```python
# Same test as above but with upper_bound
result_diff_dg = dg_1.difference_beta(dg_2, mode="upper_bound")
result_diff_dg.print_guide()
# This time root: {obj:2} is expected 
```

```python
# Case 2.2 DG(B) has paths that are not in DG(A)
# Case more docs, using lower_bound
dg_1 =  DataGuidePath()
dg_1.insert_document([doc7, doc8])
dg_2 = DataGuidePath()
dg_2.insert_document([doc7, doc8, doc9]) # doc7, doc8, doc9 all have some similar paths, but diff schema
result_diff_dg = dg_2.difference_beta(dg_1, mode="lower_bound")
result_diff_dg.print_guide() 
# We can see above that 1 doc more is found in dg2, thus ideally we want to see 1 doc in difference
# With lower_bound we guarantee in a minimum difference, so at least 1 document in dg1 is different
```

```python
# showing what each dg has from above
print("dg1:")
dg_1.print_guide()
print("dg2:")
dg_2.print_guide()
```

```python
# Case 2.2 DG(B) has paths that are not in DG(A)
# Case more docs, using upper_bound
dg_1 =  DataGuidePath()
dg_1.insert_document([doc7, doc8])
dg_2 = DataGuidePath()
dg_2.insert_document([doc7, doc8, doc9]) # doc7, doc8, doc9 all have some similar paths, but diff schema
result_diff_dg = dg_2.difference_beta(dg_1, mode="upper_bound")
result_diff_dg.print_guide() 
# With upper_bound, because paths can be in all sorts of orders, all 3 docs could be unique
# Upper_bound should output obj:2, paths and counters also appear correct 
# "root.f_name: {'int': 0, 'str': 1, 'float': 0, 'date': 0, 'obj': 0, 'arr': 0}
# root.l_name: {'int': 0, 'str': 1, 'float': 0, 'date': 0, 'obj': 0, 'arr': 0}
# root.age: {'int': 1, 'str': 0, 'float': 0, 'date': 0, 'obj': 0, 'arr': 0}
# root.sport: {'int': 0, 'str': 1, 'float': 0, 'date': 0, 'obj': 0, 'arr': 0}"
# This is because these 4 paths above can be guaranteed to appear at least that number of times uniquely,
# and thus are part of the difference.
```

```python
# same test as above but calling in reverse order
dg_1 =  DataGuidePath()
dg_1.insert_document([doc7, doc8])
dg_2 = DataGuidePath()
dg_2.insert_document([doc7, doc8, doc9]) # doc7, doc8, doc9 all have some similar paths, but diff schema
result_diff_dg = dg_1.difference_beta(dg_2, mode="lower_bound")
result_diff_dg.print_guide() 
# because dg1 is a subset fo dg2 for all documents, expected output is all 0's
```

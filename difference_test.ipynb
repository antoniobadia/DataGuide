```python
# Core tests:
# lower_bound: guaranteed minimum unique docs
# upper_bound: optimistic maximum unique docs
# Cores: root.sport:[str] , root.difficulty_lvl[str]:
# Core Test 1, upper_bound
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2])
dg_2 = DataGuidePath()
dg_2.insert_document([doc5, doc6])
result_diff_dg = dg_2.difference_beta(dg_1, mode="upper_bound")
result_diff_dg.print_guide()
# Optimistcally since dg2 contains two unique paths, we could have completely disjoint dgs
```

```python
# Core Test 1, lower_bound
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2])
dg_2 = DataGuidePath()
dg_2.insert_document([doc5, doc6])
result_diff_dg = dg_2.difference_beta(dg_1, mode="lower_bound")
result_diff_dg.print_guide()
# Accurately shows that we will have 1 unique doc with paths root.num_balls or root.field_type
```

```python
# Same tests as above but in reverse calling order, upper_bound
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2])
dg_2 = DataGuidePath()
dg_2.insert_document([doc5, doc6])
result_diff_dg = dg_1.difference_beta(dg_2, mode="upper_bound")
result_diff_dg.print_guide()
# Optimistcally since dg1 contains only 1 unique paths, we could have one unique document
# with the path root.num_balls, which is unique to dg1
# every other path appears twice, as they does in dg2
```

```python
# Same tests as above but in reverse calling order, lower_bound
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2])
dg_2 = DataGuidePath()
dg_2.insert_document([doc5, doc6])
result_diff_dg = dg_1.difference_beta(dg_2, mode="lower_bound")
result_diff_dg.print_guide()
# Accurately shows that we will have 1 unique doc with paths root.num_balls
```

```python
# Larger test, and using array in one doc
# Cores: root.sport:[str] , root.difficulty_lvl[str]:
# Core Test 2, upper_bound
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2, doc3])
dg_2 = DataGuidePath()
dg_2.insert_document([doc4, doc5, doc6, doc11])
result_diff_dg = dg_1.difference_beta(dg_2, mode="upper_bound")
result_diff_dg.print_guide()
# Accurate because both dg's have the core, 
# that only leaves root.num_balls[str] appearing twice in dg1, but once in dg1, 
# and root.involves_water[str] once in dg1 and once in dg2 max(1-1, 2-1) = 1
```

```python
# Same as above but lower_bound "guaranteed"
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2, doc3])
dg_2 = DataGuidePath()
dg_2.insert_document([doc4, doc5, doc6, doc11])
result_diff_dg = dg_1.difference_beta(dg_2, mode="lower_bound")
result_diff_dg.print_guide()
# Based on the dg's we can at most guarantee one unique document 
# because the paths in dg1 are a subset of dg2, but we have 4-3 docs =1
```

```python
# Same as above but calling in reverse, upper_bound "optimistic"
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2, doc3])
dg_2 = DataGuidePath()
dg_2.insert_document([doc4, doc5, doc6, doc11])
result_diff_dg = dg_2.difference_beta(dg_1, mode="upper_bound")
result_diff_dg.print_guide()
# There are 3 unique paths in dg2 + 2 core paths that appears more than in dg1 based on document count
# Unique paths: root.field_type[str], root.num_balls[arr], root.num_balls[str]
# the three unique paths could be seperately in doc4, doc5, and doc6 for example
# and the additional core paths would make up doc11, so max 4 documents
# the actual algorithm calculates min(total_docs=4, sum(exclusive paths=5)) = 4 because
# we cannot have more than the total docs in the dg
```

```python
# Same as above but lower_bound "guaranteed"
dg_1 =  DataGuidePath()
dg_1.insert_document([doc1, doc2, doc3])
dg_2 = DataGuidePath()
dg_2.insert_document([doc4, doc5, doc6, doc11])
result_diff_dg = dg_2.difference_beta(dg_1, mode="lower_bound")
result_diff_dg.print_guide()
# Because the paths in dg1 are a subset of dg2, at most we can have 4docs- 3docs = 1 guaranteed doc
```

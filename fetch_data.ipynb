```python
# Github_data # 1 hour of requests dg
import os
import json

folder = "github_data" # your folder name here

gh_dg = DataGuidePath()

# Loop through all .ndjson files in the folder
for filename in sorted(os.listdir(folder)):
    if not filename.endswith(".json"):
        continue

    file_path = os.path.join(folder, filename)
    print(f"Loading {file_path}...")

    # Each NDJSON file contains one JSON object per line
    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            try:
                doc = json.loads(line)
                gh_dg.insert_document(doc)
            except json.JSONDecodeError as e:
                print(f"Skipping invalid JSON line in {filename}: {e}")
              

print("\n All files loaded into DataGuidePath.")
#RUNTIME is about 08s
```

```python
# Sample code to load into pandas DF
# Importing the json data into a df for viewing data
import os
import json
import pandas as pd

# Path to your folder
folder = r"patient_data\sample-bulk-fhir-datasets-1000-patients\sample-bulk-fhir-datasets-1000-patients"

# Collect all JSON objects here
all_docs = []

# Walk through the folder and load all NDJSON files
for root, _, files in os.walk(folder):
    for fname in files:
        # Skip logs or non-ndjson/jsonl files
        if fname == "log.ndjson":
            continue
        if not fname.endswith((".ndjson", ".jsonl")):
            continue

        file_path = os.path.join(root, fname)
        print(f"Reading {file_path}...")

        # Read NDJSON line by line
        with open(file_path, "r", encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if not line:
                    continue
                try:
                    doc = json.loads(line)
                    all_docs.append(doc)
                except json.JSONDecodeError:
                    print(f"Skipping invalid JSON line in {file_path}")

# Convert to DataFrame (normalizes nested JSON)
df = pd.json_normalize(all_docs)

print(f"\n Loaded {len(df)} total rows from {len(all_docs)} JSON objects.")
```

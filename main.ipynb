```python
#Testing union, difference, and intersect methods
#Data
data1 = {"a": 2, "b": {"c": 7, "d": 'bar', "e": 4}, "f": [1, 2, 3]}
data2 = {"a": 2, "b": {"c": 'bar'}, "g": 7}

d_guide1 = DataGuide()
d_guide2 = DataGuide()

d_guide1.insert_document(data1)
d_guide2.insert_document(data2)
```

```python
union_d_guide1_d_guide2 = d_guide1.union(d_guide2)
union_d_guide1_d_guide2.print_guide()
#Union appears to be working well, as shown below, paths like b.c or f.* are found
```

```python
# Creating data to test difference
data3 = [
    { "a": 1, "common": { "x": 10 }, "u1": "foo" },
    { "a": 2, "common": { "x": "twenty" }, "u2": "bar" },
    { "a": 3, "common": { "x": 30 } }
]

data4 = [
    { "a": 100, "common": { "x": 40 }, "v1": 3.14 },
    { "a": 200, "common": { "x": 20 }},
    { "a": 300, "common": { "x": "gamma" }, "v2": 2.71 }
]

d_guide3 = DataGuide()
d_guide4 = DataGuide()
```

```python
# This will show the difference in data guides
d_guide3.insert_document(data3)
d_guide4.insert_document(data4)

difference_d_guide3_d_guide4 = d_guide3.difference(d_guide4)
difference_d_guide3_d_guide4.print_guide()
```

```python
#This will show where the data guides intersect
intersect_d_guide3_d_guide4 = d_guide3.intersect(d_guide4)
intersect_d_guide3_d_guide4.print_guide()
```

```python
print(intersect_d_guide3_d_guide4.total_docs)
```
